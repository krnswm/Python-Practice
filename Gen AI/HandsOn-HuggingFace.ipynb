## Performing the Task

### **Sentiment Analysis** - Finding emotions or opinion from the text

- **Pipeline is a step-by-step process that automatically handles all steps from input to output using a pre trained model.**

- **It is like a short-cut which combines each and every step in a single line**

## Importing Pipline

```python
from transformers import pipeline
```

**Using this pipeline fro sentiment analysis**

```python
model = pipeline("sentiment-analysis")
```

```python
result = model(' i am learning Gen AI and i am very happy')
```

```python
print(result)
```

```python
s2 = model(' i am not in a good mood today!!')
print(s2)
```

```python
s2[0]['label']
```

#### NOW WE ARE SELECTING ONE SPECIFIC FINANCIAL SENTIMENT ANALYSIS MODEL

```python
from transformers import AutoTokenizer , AutoModelForSequenceClassification, pipeline
```

```python
tokenizer = AutoTokenizer.from_pretrained('mrm8488/distilroberta-finetuned-financial-news-sentiment-analysis')
```

```python
model = AutoModelForSequenceClassification.from_pretrained('mrm8488/distilroberta-finetuned-financial-news-sentiment-analysis')
```

```python
financial_analysis = pipeline('sentiment-analysis',model=model , tokenizer=tokenizer)
```

```python
text = 'XYZ coorporations stock soared by 20% after reporting record-breaking annual profits'
```

```python
ans = financial_analysis(text)
print(ans)
```

**Irrelevant information**

```python
t1 = 'i am going on a trip'
ans1 = financial_analysis(t1)
print(ans1)
```

# **Another model**

```python
from transformers import AutoTokenizer , AutoModelForSequenceClassification, pipeline
```

```python
tokenize = AutoTokenizer.from_pretrained('nlptown/bert-base-multilingual-uncased-sentiment')
```

```python
nlp = AutoModelForSequenceClassification.from_pretrained('nlptown/bert-base-multilingual-uncased-sentiment')
```

```python
ans = pipeline('sentiment-analysis',model=nlp , tokenizer=tokenize)
```

```python
text = 'wow , another superhero movie bombed at the box-office , so -surprising'
res = ans(text)
print(res)
```

## **TEXT-GENERATION**

```python
# Importing pipelne from transformers
from transformers import pipeline
```

```python
generator = pipeline("text-generation",model="gpt2")
```

```python
prompt = "Once upon a time in a forest,"
# generate text from the prompt
response = generator(prompt,max_length=50,num_return_sequences=1)

# print the generated text
print(response[0]['generated_text'])
```
